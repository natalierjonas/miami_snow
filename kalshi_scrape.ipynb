```python
import requests
import pandas as pd

# API endpoint
BASE_URL = "https://api.elections.kalshi.com/v1/social/trades"

# Parameters
params = {
    "series_ticker": "KXMIASNOWM",  # your series
    "page_size": 100                 # max per request
}

all_trades = []

while True:
    r = requests.get(BASE_URL, params=params)
    data = r.json()

    trades = data.get("trades", [])
    all_trades.extend(trades)

    print(f"Fetched {len(trades)} trades")

    # Pagination cursor
    cursor = data.get("cursor")
    if not cursor:
        break
    params["cursor"] = cursor

print(f"Total trades fetched: {len(all_trades)}")

# Convert to DataFrame
df = pd.DataFrame(all_trades)

# Optional: parse timestamp to readable datetime
if "created_at" in df.columns:
    df["created_at"] = pd.to_datetime(df["created_at"])

# Save to CSV
df.to_csv("kalshi_activity_full.csv", index=False)
print("✅ Saved CSV as 'kalshi_activity_full.csv'")
```
